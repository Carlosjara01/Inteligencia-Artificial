Instalar e importar librerías

```python
# Librerías principales para trabajar con datos
import pandas as pd
import numpy as np

# Librerías para normalización
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Librería para descargar el archivo limpio
from google.colab import files
```

Cargar el dataset

```python
# Cambia el nombre si tu archivo tiene otro nombre
nombre_archivo = "sales_data_sample.csv"

# Cargar el dataset
# encoding="latin1" ayuda cuando el CSV tiene caracteres especiales
df = pd.read_csv(nombre_archivo, encoding="latin1")

# Mostrar las primeras filas
df.head()
```

Revisar tamaño y estructura

```python
# Cantidad de filas y columnas
print("Filas y columnas:", df.shape)

# Mostrar nombres de columnas
print("\nColumnas del dataset:")
print(df.columns.tolist())

# Ver tipos de datos
print("\nTipos de datos:")
print(df.dtypes)
```

Revisar valores nulos

```python
# Ver cantidad de valores nulos por columna
print("Valores nulos por columna:")
print(df.isnull().sum())
```

Crear una copia del dataset

```python
# Creamos una copia para trabajar sin dañar el dataset original
df_limpio = df.copy()
```

Limpiar espacios en textos

```python
# Elimina espacios innecesarios en columnas de texto
for columna in df_limpio.select_dtypes(include=["object"]).columns:
    df_limpio[columna] = df_limpio[columna].astype(str).str.strip()
```

Convertir fechas si existen

```python
# Intentamos convertir columnas que tengan la palabra DATE en formato fecha
for columna in df_limpio.columns:
    if "DATE" in columna.upper():
        df_limpio[columna] = pd.to_datetime(df_limpio[columna], errors="coerce")
        print(f"Columna convertida a fecha: {columna}")
```

Crear columnas útiles desde fechas

```python
# Si existe ORDERDATE, creamos columnas de año, mes y día
if "ORDERDATE" in df_limpio.columns:
    df_limpio["ORDER_YEAR"] = df_limpio["ORDERDATE"].dt.year
    df_limpio["ORDER_MONTH"] = df_limpio["ORDERDATE"].dt.month
    df_limpio["ORDER_DAY"] = df_limpio["ORDERDATE"].dt.day
```

Rellenar valores nulos

```python
# Separar columnas numéricas y categóricas
columnas_numericas = df_limpio.select_dtypes(include=["int64", "float64"]).columns
columnas_categoricas = df_limpio.select_dtypes(include=["object"]).columns

# Rellenar nulos numéricos con el promedio
for columna in columnas_numericas:
    df_limpio[columna] = df_limpio[columna].fillna(df_limpio[columna].mean())

# Rellenar nulos categóricos con "Desconocido"
for columna in columnas_categoricas:
    df_limpio[columna] = df_limpio[columna].fillna("Desconocido")

# Verificar nuevamente nulos
df_limpio.isnull().sum()
```

Seleccionar columnas numéricas para normalizar

```python
# Volvemos a detectar columnas numéricas después de la limpieza
columnas_numericas = df_limpio.select_dtypes(include=["int64", "float64"]).columns

print("Columnas numéricas que se pueden normalizar:")
print(columnas_numericas.tolist())
```

. Normalización con StandardScaler

```python
# StandardScaler transforma los datos para que tengan:
# media cercana a 0 y desviación estándar cercana a 1

scaler = StandardScaler()

# Creamos una copia para no perder los valores originales
df_normalizado = df_limpio.copy()

# Aplicamos normalización solo a columnas numéricas
df_normalizado[columnas_numericas] = scaler.fit_transform(df_limpio[columnas_numericas])

# Mostrar resultado
df_normalizado.head()
```

Guardar dataset normalizado

```python
# Guardar el dataset limpio y normalizado
nombre_salida = "dataset_limpio_normalizado.csv"

df_normalizado.to_csv(nombre_salida, index=False, encoding="utf-8-sig")

print("Dataset guardado correctamente:", nombre_salida)
```
